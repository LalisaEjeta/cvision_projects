```python
from fast_alpr import ALPR

alpr = ALPR()

results = alpr.predict("car.jpg")

for r in results:
    if r.ocr:
        print(f"Plate: {r.ocr.text}")
        print(f"Detection confidence: {r.detection.confidence:.4f}")
        print(f"Region: {r.ocr.region}")
        print(f"Region confidence: {r.ocr.region_confidence:.4f}")
```
