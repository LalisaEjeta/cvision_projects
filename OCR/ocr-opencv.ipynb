# First image

```python
import cv2
import pytesseract

img = cv2.imread("text-image.png")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# upscale
gray = cv2.resize(
    gray,
    None,
    fx=2,
    fy=2,
    interpolation=cv2.INTER_CUBIC
)

# adaptive threshold
gray = cv2.adaptiveThreshold(
    gray,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

text = pytesseract.image_to_string(
    gray,
    config="--psm 6"
)

print(repr(text))
```
